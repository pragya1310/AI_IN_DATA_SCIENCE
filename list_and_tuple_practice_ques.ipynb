### 1. Find the largest and smallest element in a list.

```python
# Create a list of numbers
lst = [10, 25, 5, 40, 15, 2]

# Assume the first element is both largest and smallest
largest = lst[0]
smallest = lst[0]

# Check every element in the list
for num in lst:

    # Check if current number is greater than largest
    if num > largest:
        largest = num

    # Check if current number is smaller than smallest
    if num < smallest:
        smallest = num

# Print the results
print("Largest element:", largest)
print("Smallest element:", smallest)
```

### 2. Remove duplicates from a list.

```python
# Create a list with duplicate elements
lst = [10, 20, 10, 30, 20, 40, 30]

# Create an empty list to store unique elements
unique = []

# Check each element in the list
for num in lst:

    # Add the element only if it is not already present
    if num not in unique:
        unique.append(num)

# Print the list after removing duplicates
print("Original list:", lst)
print("List after removing duplicates:", unique)
```

### 3. Reverse a list.

```python
# Create a list
lst = [10, 20, 30, 40, 50]

# Reverse the list using slicing
reverse_lst = lst[::-1]

# Print the original list
print("Original list:", lst)

# Print the reversed list
print("Reversed list:", reverse_lst)
```

### 4. Count even and odd numbers in a list.

```python
# Create a list of numbers
lst = [10, 15, 20, 25, 30, 35, 40]

# Create counters for even and odd numbers
even_count = 0
odd_count = 0

# Check each number in the list
for num in lst:

    # Check if the number is even
    if num % 2 == 0:
        even_count = even_count + 1

    # Otherwise, the number is odd
    else:
        odd_count = odd_count + 1

# Print the results
print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
```

### 5. Merge and sort two lists.

```python
# Create the first list
list1 = [10, 30, 20, 50]

# Create the second list
list2 = [40, 15, 5, 25]

# Merge both lists
merged_list = list1 + list2

# Sort the merged list
merged_list.sort()

# Print the result
print("Merged and sorted list:", merged_list)
```

### 6. Find the second largest element in a list.

```python
# Create a list of numbers
lst = [10, 25, 40, 15, 30, 40]

# Remove duplicate elements
unique = []

for num in lst:
    if num not in unique:
        unique.append(num)

# Find the largest and second largest
largest = unique[0]
second_largest = unique[0]

for num in unique:

    # If current number is greater than largest
    if num > largest:
        second_largest = largest
        largest = num

    # If current number is between largest and second largest
    elif num > second_largest:
        second_largest = num

# Print the result
print("Largest element:", largest)
print("Second largest element:", second_largest)
```

### 7. Check if an element exists in a tuple.

```python
# Create a tuple
tup = (10, 20, 30, 40, 50)

# Take an element from the user
num = int(input("Enter the number to search: "))

# Check if the element exists in the tuple
if num in tup:
    print("Element exists in the tuple")
else:
    print("Element does not exist in the tuple")
```

### 8. Count the occurrences of an element in a tuple.

```python
# Create a tuple
tup = (10, 20, 10, 30, 10, 40, 20)

# Take an element from the user
num = int(input("Enter the number: "))

# Count how many times the element occurs
count = tup.count(num)

# Print the result
print("Number of occurrences:", count)
```

### 9. Sort a list of tuples by the second item.

```python
# Create a list of tuples
students = [
    ("pragya", 85),
    ("Payal", 92),
    ("Amar", 78),
    ("Niharika", 88)
]

# Sort the list by the second item of each tuple
students.sort(key=lambda x: x[1])

# Print the sorted list
print("Sorted list:", students)
```

### 10. Convert a tuple to a list and a list to a tuple.

```python
# Create a tuple
tup = (10, 20, 30, 40)

# Convert tuple into a list
lst = list(tup)

# Create a list
numbers = [50, 60, 70, 80]

# Convert list into a tuple
new_tup = tuple(numbers)

# Print the results
print("Original tuple:", tup)
print("Tuple converted to list:", lst)

print("Original list:", numbers)
print("List converted to tuple:", new_tup)
```
