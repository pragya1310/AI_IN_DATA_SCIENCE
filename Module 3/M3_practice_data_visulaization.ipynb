# WAP to  plot a bar chart of subject with respect to score_2026,score_2025,score_2024

```python
import matplotlib.pyplot as plt
```

```python
subject=["math","english" , "science","hindi","computer"]
```

```python
score_2026=[78,89,80,92,90]
score_2025=[87,98,88,88,75]
score_2024=[65,71,80,75,68]
```

```python
x=range(len(subject))
```

# Write a program to create chart of units,profits and months of a company

```python
import matplotlib.pyplot as plt
```

```python
#create the data
months=["January","February","March","April","May","June","July","August","September","October","November","Decemeber"]

units=[ 100,45,67,89,54,63,98,34,56,33,23,85]

profits=[100000,4000,6000,8000,5000,7400,9200,2800,5800,7800,9800,4500]

```

#Line Chart

for the line chart months and profits

```python
#plotting line chart

plt.figure(figsize=(15,5))
plt.plot(months , profits,marker='s')

#title and label
plt.title("monthly profit")
plt.xlabel("months")
plt.ylabel("profits")

# show line plot
plt.show()
```

## Bar Chart

```python
# figure resize
plt.figure(figsize=(15,5))

# plotting bar graph
plt.bar(months,units)

# title and label
plt.title("months vs units",color='red')
plt.xlabel("months",color='red')
plt.ylabel("units",color='red')


# show the chart
plt.show()
```

# Histogram

```python
plt.hist(units, bins=5,edgecolor='r')
plt.xlabel("units")
plt.ylabel("Frequency")
plt.title("Units Distribution")
plt.show()
```

# Scatter plot

```python
# scatter plotting

plt.scatter(units,profits,color='red')

#title and label
plt.title("units vs profits")
plt.xlabel("units")
plt.ylabel("profits")

#show the chart
plt.show()
```
