for loop with range()

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

for i in range(len(x)):
    print(x[i], y[i])

plt.plot(x, y, marker='o')
plt.show()
```

for loop with enumerate()

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

for index, value in enumerate(y):
    print("Index:", index, "Value:", value)

plt.plot(x, y, marker='o')
plt.show()
```

while loop

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

i = 0

while i < len(x):
    print(x[i], y[i])
    i += 1

plt.plot(x, y, marker='o')
plt.show()
```

range() → generates numbers
index → position of an element
enumerate() → gives index + value
for → repeats for a sequence
while → repeats while a condition is true

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

for i in range(len(x)):
  plt.plot(x, y, marker='o')
  plt.text(x[i], y[i], str([x[i],y[i]]))
plt.show()
```

# 3.3 Bar Chart

```python
# create  the data

subject=["python","java","DSA","c++"]
score=[40,60,65,85]

# plot the bar graph
plt.bar(subject,score)

#create title and label
plt.title("subject vs score")
plt.xlabel("subject")
plt.ylabel("score")

# show plot
plt.show()
```

```python
# color and grid
plt.bar(subject,score ,color=['red','green','blue','pink'])

#create title and label
plt.title("subject vs score")
plt.xlabel("subject")
plt.ylabel("score")

# grid
plt.grid(axis='y')  # y for

# show plot
plt.show()
```

# 3.4 Pie Chart

used for data distribution like in 100 marks it shows subject wise score in a circle

```python
# create the data
subject=["technical","aptitude"," gk","english"]
score=[50,20,20,10]

#pie chart
'plt.pie(score,labels=subject)'

# percentage
plt.pie(score,labels=subject,autopct='%1.1f%%',startangle=170,explode=(0,0,0,0.5) ) #1.1 used to show decimal places

'''"startangle used to rotate the pie chart"
explode=(0,0,0,0.5) used to highlight the major part or pie chart'''

#title
plt.title("marks distribution")

#plot show
plt.show()
```

```python

# create the data
subject=["technical","aptitude"," gk","english"]
score=[50,20,20,10]
colors=['red','green','blue','grey'] # Adjusted colors list to match score length

#pie chart
'plt.pie(score,labels=subject)'

# percentage
# Pass 'score' as the data, and 'colors' using the 'colors' keyword argument.
# Added 'explode' to highlight the last segment, similar to the previous cell.
plt.pie(score,labels=subject,autopct='%1.1f%%',startangle=170, colors=colors,shadow=True, explode=[0, 0, 0, 0.5])  #1.1 used to show decimal places

"startangle used to rotate the pie chart"


#title
plt.title("marks distribution")

#legend
'plt.legend()'
'plt.legend(subject)'
plt.legend(score)
'plt.legend(subject,loc="upper right")'

#plot show
plt.show()
```
