# 1. Series

```python
import pandas as pd    # alies  (as)

```

```python
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks)
print(ds)
print(type(ds))
```

```python
#customized index
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks,index=['a','b','c','d','e','f'])


print(ds)
```

```python
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks,index=['a','b','c','d','e'])  #  Length of values (6) does not match length of index (5)


print(ds)
```

```python
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks,index=['a','b','c','d','e','f','g'])#ValueError: Length of values (6) does not match length of index (7)


print(ds)
```

```python
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks,index=['a','b','c','d','e','f'],dtype=float)


print(ds)
```

```python
# add file name
Marks=[90,91,92,93,94,94]
ds=pd.Series(Marks,index=['a','b','c','d','e','f'],dtype=float,name="marksheet")


print(ds)
```

# Handling missing values

```python
# create a dict. using np.NaN
#
import numpy as np
student={
    'Name':['Pragya','Payal','Niharika','Chhavi','Sarthak'],
    'Roll no.':[70,67,62,34,np.nan],
    'Marks':[92,93,np.nan,98,99],
    'Subject':['Chem',np.nan,'BIO',np.nan,'physics'],
    'City':['noida','noida','siwan',np.nan,np.nan]
}

df=pd.DataFrame(student)
df
```

```python
df.info()
```

```python
df.isnull()
#non null=false and null=true
```

```python
df.isnull().sum()
```

```python
print(df.isnull().sum().sum())
```

```python
df.dropna()
```

```python
df['Marks']=df['Marks'].fillna(0)
```

```python
df.dropna(subset=['Roll no.'])
```

```python
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
print(df['Marks'])
```

```python
df['Name'].fillna("unknown")
```
