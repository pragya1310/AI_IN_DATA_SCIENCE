Question 3: Count Words, Lines, and Characters
Problem Statement:
Write a Python program that reads a text file and displays:
 Total number of lines
 Total number of words
 Total number of characters (including spaces)

```python
# Open the file in read mode
file = open("student.txt", "r")

# Read the entire content
data = file.read()

# Count the number of lines
lines = data.split("\n")

# Count the number of words
words = data.split()

# Count the number of characters
characters = len(data)

# Display the results
print("Total lines:", len(lines))
print("Total words:", len(words))
print("Total characters:", characters)

# Close the file
file.close()
```

Question 4: Copy File Content
Problem Statement:
Write a Python program to:
 Read data from source.txt.
 Copy all the contents into destination.txt.
 Display a success message after copying.
Expected Concepts:
 Reading one file
 Writing another file
 r and w mode

```python
# Create source.txt and write some data
source = open("source.txt", "w")

source.write("Hello, my name is Rahul.\n")
source.write("I am studying B.Tech.")

source.close()


# Open source.txt in read mode
source = open("source.txt", "r")

# Read the content
data = source.read()

source.close()


# Open destination.txt in write mode
destination = open("destination.txt", "w")

# Copy the content
destination.write(data)

destination.close()


# Display success message
print("File copied successfully")
```

Question 5: Student Record Management System
Problem Statement:
Create a menu-driven Student Record Management System using file handling.
Menu
===== Student Record Management =====
1. Add Student
2. View All Students
3. Search Student by Name
4. Exit
Requirements
 Store student records in students.txt.
 Each record should contain:
o Name
o Roll Number
o Course
 Allow users to add multiple records.
 Display all records.
 Search for a student by name.
 Continue showing the menu until the user chooses Exit.

```python
while True:

    # Display menu
    print("\n===== Student Record Management =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student by Name")
    print("4. Exit")

    # Take choice from user
    choice = input("Enter your choice: ")

    # -------------------------------
    # 1. Add Student
    # -------------------------------
    if choice == "1":

        name = input("Enter student name: ")
        roll = input("Enter roll number: ")
        course = input("Enter course: ")

        # Open file in append mode
        file = open("students.txt", "a")

        # Store student details
        file.write(name + "," + roll + "," + course + "\n")

        # Close file
        file.close()

        print("Student added successfully!")

    # -------------------------------
    # 2. View All Students
    # -------------------------------
    elif choice == "2":

        file = open("students.txt", "r")

        print("\n----- All Students -----")

        # Read each record
        for line in file:
            data = line.strip().split(",")

            print("Name:", data[0])
            print("Roll Number:", data[1])
            print("Course:", data[2])
            print("----------------------")

        file.close()

    # -------------------------------
    # 3. Search Student by Name
    # -------------------------------
    elif choice == "3":

        search_name = input("Enter student name to search: ")

        file = open("students.txt", "r")

        found = False

        # Read each record
        for line in file:

            data = line.strip().split(",")

            # Compare names
            if data[0].lower() == search_name.lower():

                print("\nStudent Found!")
                print("Name:", data[0])
                print("Roll Number:", data[1])
                print("Course:", data[2])

                found = True

        file.close()

        # If student was not found
        if found == False:
            print("Student not found.")

    # -------------------------------
    # 4. Exit
    # -------------------------------
    elif choice == "4":

        print("Thank you!")
        break

    # Invalid choice
    else:
        print("Invalid choice. Please try again.")
```

Question: Remove Duplicate Lines from a File
Write a Python program to:
 Read all lines from a text file.
 Remove duplicate lines while preserving the original order.
 Save the unique lines into a new file named unique.txt.

```python
# Open the file in read mode
file = open("student.txt", "r")

# Read all lines
lines = file.readlines()

# Close the file
file.close()

# Create an empty list
unique_lines = []

# Remove duplicate lines
for line in lines:
    if line not in unique_lines:
        unique_lines.append(line)

# Open unique.txt in write mode
file = open("unique.txt", "w")

# Write unique lines into the file
for line in unique_lines:
    file.write(line)

# Close the file
file.close()

# Display the unique lines
print("Unique lines are:")

for line in unique_lines:
    print(line, end="")

print("\nUnique file created successfully!")
```
