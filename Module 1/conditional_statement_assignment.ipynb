```python
# 1. Program to check whether a year is a leap year or not.
year=int(input("enter the year"))

if year%4==0 and year%100 !=0:
  print("leap year")
elif year%400==0:
  print("leap year")
else:
  print("not leap year")
```

```python
# 3. Program to check whether a character is an uppercase letter, lowercase letter, digit, or special character.

chars=input("enter the char ")
if 'a'<= chars <='z':
  print("lowercase")
elif 'A' <= chars <= 'Z':
  print("uppercase")
elif '0'<= chars <='9':
  print("digit")
else:
  print("special char")
```

```python
# 4. Program to calculate electricity bill using different unit slabs.
units=int(input("enter the units:  "))
if units<=50:
  bill= units*.5
  print(bill)
elif units <=150:
  bill=(50*.5)+ (units-50)*0.75
  print(bill)
elif units <=250:
  bill= (50*.5)+(100*.75)+(units-150)*1
  print(bill)
else:
  bill = (50*.5)+(100*.75)+(150*1)+(units-250)*1.25
  print(bill)
```

```python
# 5. Program to determine whether a triangle is Equilateral, Isosceles, or Scalene.
a=int(input("enter the side a "))
b=int(input("enter the side b "))
c=int(input("enter the side c "))

if a==b==c:
  print("equilateral")
elif a==b or a==c or b==c:
  print("isosceles")
else:
  print("scalene")
```

```python
from IPython.core.compilerop import operator
# 6. Program to create a simple calculator using if-elif-else.
num = float(input("Enter a number: "))
num1 = float(input("Enter a number: "))
operator=input("enter the operator")
if operator=='+':
  print(num+num1)
elif operator=='-':
  print(num-num1)
elif operator=='*':
  print(num*num1)
elif operator == '/':
  print(num/num1)
elif operator== '**':
  print(num**num1)
else:
  print("error")

```

```python
# 7. Program to calculate income tax according to salary ranges.
'''₹8,00,001 to ₹12,00,000: 10%
₹12,00,001 to ₹16,00,000: 15%
₹16,00,001 to ₹20,00,000: 20%
₹20,00,001 to ₹24,00,000: 25%
Above ₹24,00,000: 30%'''
salary = float(input("enter the salary"))

if salary<=400000:
  print("no tax")
elif salary <= 800000:
  salary_inhand= (salary-salary*0.05)
  print(salary_inhand)
elif salary <= 1200000:
  salary_inhand= (salary-salary*0.10)
  print(salary_inhand)
elif salary <= 1600000:
  salary_inhand = (salary -salary*0.15)
  print(salary_inhand)
elif salary <=2000000:
  salary_inhand= (salary-salary*0.20)
  print(salary_inhand)
elif salary <=2400000:
  salary_inhand= (salary-salary*0.25)
  print(salary_inhand)
else:
  salary_inhand= (salary-salary*0.30)
  print(salary_inhand)
```

```python
# 9. Program to determine whether a point lies in First quadrant, Second quadrant, Third quadrant, Fourth quadrant, On axis, or At origin.
x=int(input("enter the x "))
y=int(input("enter the y "))

if x==0 and y==0:
  print("at origin")
elif x==0 and y>0:
  print("on y axis")
elif x>0 and y ==0:
  print("on x axis")
elif x>0 and y>0:
  print("first quadrant")
elif x<0 and y>0:
  print("second quadrant")
elif x<0 and y<0:
  print("third quadrant")
else:
  print("fourth quadrant")
```

```python
 #10. Program to assign grades based on marks and display distinction for high scores.
print("\n--- Grade Assignment ---")
marks = int(input("Enter the marks: "))
if 90 <= marks <= 100:
    print(f"Marks {marks}: A+ ")
elif 80 <= marks < 90:
    print(f"Marks {marks}: A")
elif 70 <= marks < 80:
    print(f"Marks {marks}: B")
elif 60 <= marks < 70:
    print(f"Marks {marks}: C")
elif 50 <= marks < 60:
    print(f"Marks {marks}: D")
elif 0 <= marks < 50:
    print(f"Marks {marks}: Fail")
else:
    print(f"Marks {marks}: Invalid Marks")
```
