```python
#Write a function that takes a number and returns its factorial.


# Function to find factorial
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


# Take input
num = int(input("Enter a number: "))

# Call function
result = factorial(num)

print("Factorial:", result)
```

```python
#Write a function to check whether a number is prime or not.

# Function to check prime number
def prime(num):

    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


# Take input
num = int(input("Enter a number: "))

# Check result
if prime(num):
    print("Prime number")
else:
    print("Not a prime number")
```

```python
#Write a function that takes three numbers and returns the largest.

# Function to find largest number
def largest(a, b, c):

    if a > b and a > c:
        return a
    elif b > c:
        return b
    else:
        return c


# Take input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

# Call function
result = largest(a, b, c)

print("Largest:", result)
```

```python
# Function to check palindrome
def palindrome(text):

    reverse = ""

    for ch in text:
        reverse = ch + reverse

    if text == reverse:
        return True
    else:
        return False


# Take input
text = input("Enter a string: ")

# Check result
if palindrome(text):
    print("Palindrome")
else:
    print("Not a palindrome")
```
