1. Armstrong Number in a Range
Print all Armstrong numbers between 1 and N.

```python
# Take the limit from the user
n = int(input("Enter the value of N: "))

# Check every number from 1 to N
for num in range(1, n + 1):

    # Store the original number
    original = num

    # Find the number of digits
    digits = len(str(num))

    # Variable to store the sum
    total = 0

    # Make a copy of the number
    temp = num

    # Calculate the sum of powers of digits
    while temp > 0:
        digit = temp % 10
        total = total + digit ** digits
        temp = temp // 10

    # Check if the number is Armstrong
    if total == original:
        print(original)
```

2. Happy Number
A number is called Happy if repeatedly replacing it with the sum of the squares of its digits
eventually reaches 1.

```python
# Take a number from the user
num = int(input("Enter a number: "))

# Store the original number
original = num

# Repeat until the number becomes 1
while num != 1 and num != 4:

    # Variable to store the sum
    total = 0

    # Find the sum of squares of digits
    while num > 0:
        digit = num % 10
        total = total + digit ** 2
        num = num // 10

    # Store the new value
    num = total

# Check whether the number is happy
if num == 1:
    print(original, "is a Happy Number")
else:
    print(original, "is not a Happy Number")
```

3. Find the Largest Digit in a Number

```python
# Take a number from the user
num = int(input("Enter a number: "))

# Store the largest digit
largest = 0

# Check each digit
while num > 0:

    # Get the last digit
    digit = num % 10

    # Check if the digit is greater than largest
    if digit > largest:
        largest = digit

    # Remove the last digit
    num = num // 10

# Print the largest digit
print("Largest digit:", largest)
```

4. Password Strength Checker (Using Loops)
Check whether a password contains:
 Uppercase  
 Lowercase  
 Digit  
 Special Character  
 Minimum 8 characters

```python
# Take password from the user
password = input("Enter your password: ")

# Create variables to check each condition
uppercase = False
lowercase = False
digit = False
special = False

# Check every character of the password
for ch in password:

    # Check for uppercase letter
    if ch.isupper():
        uppercase = True

    # Check for lowercase letter
    elif ch.islower():
        lowercase = True

    # Check for digit
    elif ch.isdigit():
        digit = True

    # Otherwise, it is a special character
    else:
        special = True

# Check all conditions
if len(password) >= 8 and uppercase and lowercase and digit and special:
    print("Strong Password")
else:
    print("Weak Password")
```
