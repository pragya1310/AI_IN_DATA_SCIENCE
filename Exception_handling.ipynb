Question 1: Handle Division by Zero

```python
# Take two numbers from the user
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

try:
    # Try to divide the numbers
    result = a / b

    print(result)

except ZeroDivisionError:
    # Handle division by zero
    print("Cannot divide by zero")
```

Question 2: Handle Multiple Exceptions

```python
try:
    # Take input
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    # Divide
    result = a / b

    print("Result:", result)

except ValueError:
    # Handles non-integer input
    print("Please enter numbers only")

except ZeroDivisionError:
    # Handles division by zero
    print("Cannot divide by zero")
```

Question 3: Try-Except-Else

```python
try:
    # Take input
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    # Divide
    result = a / b

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Please enter numbers only")

else:
    # Runs only when there is no error
    print("Result:", result)
```

Question 4: Finally Block

```python
try:
    num = int(input("Enter a number: "))

    print("Number:", num)

except ValueError:
    print("Invalid input")

finally:
    # This always executes
    print("Program completed")
```

File Handling + Exception Handling

```python
try:
    # Open the file
    file = open("student.txt", "r")

    # Read the content
    data = file.read()

    # Display the content
    print(data)

    # Close the file
    file.close()

except FileNotFoundError:
    # Handle missing file
    print("File not found")
```

Question: Use raise to Check Age

```python
try:
    # Take age from the user
    age = int(input("Enter your age: "))

    # Check age
    if age < 18:
        raise ValueError("You are not eligible to vote")

    print("You are eligible to vote")

except ValueError as e:
    # Handle the error
    print("Error:", e)
```
