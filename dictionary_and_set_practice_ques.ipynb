3. Write a program to sort a dictionary by its values.


```python
# Create a dictionary
students = {
    "Riya": 85,
    "Priti": 92,
    "Amar": 78,
    "Nihu": 88
}

# Sort the dictionary by values
sorted_students = dict(sorted(students.items(), key=lambda x: x[1]))

# Print the sorted dictionary
print("Dictionary sorted by values:")
print(sorted_students)
```

4. Given two sets, check if one set is a subset of another.


```python
# Create two sets
set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}

# Check if set1 is a subset of set2
if set1.issubset(set2):
    print("Set1 is a subset of Set2")
else:
    print("Set1 is not a subset of Set2")
```

5. Write a program to check whether two lists have at least one common element using sets

```python
# Create the first list
list1 = [10, 20, 30, 40]

# Create the second list
list2 = [50, 60, 30, 70]

# Convert both lists into sets
set1 = set(list1)
set2 = set(list2)

# Find common elements using intersection
common = set1.intersection(set2)

# Check if there is at least one common element
if common:
    print("The lists have at least one common element")
    print("Common elements:", common)
else:
    print("The lists do not have any common element")
```
